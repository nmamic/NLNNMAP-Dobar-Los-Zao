```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report


# 1. Učitavanje i priprema MNIST podataka (s labelama)
(x_train, y_train), (x_test, y_test) = mnist.load_data()

x_train = x_train.astype("float32") / 255.0
x_train = np.expand_dims(x_train, axis=-1)

x_test = x_test.astype("float32") / 255.0
x_test = np.expand_dims(x_test, axis=-1)

print("Train shape:", x_train.shape, "Test shape:", x_test.shape)

EPOCHS = 15
BATCH_SIZE = 128


# 2. Arhitektura mreže
input_img = layers.Input(shape=(28, 28, 1))

# ----- ENKODER: konvolucijski dio -----
x = layers.Conv2D(3, (3, 3), strides=2, activation="relu", padding="same")(input_img)   # 14x14x3
x = layers.Conv2D(3, (3, 3), strides=2, activation="relu", padding="same")(x)            # 7x7x3

# ----- ENKODER: fully connected dio -----
shape_before_flatten = x.shape[1:]  # (7, 7, 3) - treba nam za reshape u dekoderu
x = layers.Flatten()(x)                                  # 7*7*3 = 147
x = layers.Dense(96, activation="relu")(x)                # 147 -> 96
encoded = layers.Dense(64, activation="relu", name="central_layer")(x)  # 96 -> 64 (BOTTLENECK)

# ----- DEKODER: fully connected dio -----
x = layers.Dense(96, activation="relu")(encoded)          # 64 -> 96
x = layers.Dense(147, activation="relu")(x)               # 96 -> 147
x = layers.Reshape(shape_before_flatten)(x)               # 147 -> 7x7x3

# ----- DEKODER: konvolucijski (dekonvolucijski) dio -----
x = layers.Conv2DTranspose(3, (3, 3), strides=2, activation="relu", padding="same")(x)   # 7x7x3 -> 14x14x3
x = layers.Conv2DTranspose(3, (3, 3), strides=2, activation="relu", padding="same")(x)   # 14x14x3 -> 28x28x3
decoded = layers.Conv2D(1, (3, 3), activation="sigmoid", padding="same")(x)              # 28x28x1 (izlazna slika)

autoencoder = models.Model(input_img, decoded, name="AE_Conv_FC_slika")
encoder = models.Model(input_img, encoded, name="Encoder_central64")
autoencoder.compile(optimizer="adam", loss="binary_crossentropy")

autoencoder.summary()



# 3. Treniranje modela
history = autoencoder.fit(
    x_train, x_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    shuffle=True,
    validation_data=(x_test, x_test),
    verbose=1
)



# 4. Graf gubitka (loss) kroz epohe
plt.figure(figsize=(6, 4))
plt.plot(history.history["loss"], label="train loss")
plt.plot(history.history["val_loss"], label="val loss")
plt.title("Gubitak kroz epohe - arhitektura sa slike (Central Layer = 64)")
plt.xlabel("Epoha")
plt.ylabel("Binary Crossentropy")
plt.legend()
plt.tight_layout()
plt.show()



# 5. Vizualizacija rekonstrukcija (original vs. rekonstrukcija)
decoded_imgs = autoencoder.predict(x_test)

n = 10
plt.figure(figsize=(20, 4))
plt.suptitle("Rekonstrukcije - arhitektura sa slike (Conv1->Conv2->FC->Central(64)->FC->Deconv1->Deconv2)", fontsize=12)
for i in range(n):
    ax = plt.subplot(2, n, i + 1)
    plt.imshow(x_test[i].reshape(28, 28), cmap="gray")
    plt.title("Original")
    ax.get_xaxis().set_visible(False)
    ax.get_yaxis().set_visible(False)

    ax = plt.subplot(2, n, i + 1 + n)
    plt.imshow(decoded_imgs[i].reshape(28, 28), cmap="gray")
    plt.title("Rekonstrukcija")
    ax.get_xaxis().set_visible(False)
    ax.get_yaxis().set_visible(False)
plt.tight_layout()
plt.show()


# 6. Priprema podataka za SVM (uzorkovanje radi brzine)
N_TRAIN_SVM = 5000
N_TEST_SVM = 2000

rng = np.random.RandomState(42)
train_idx = rng.choice(len(x_train), N_TRAIN_SVM, replace=False)
test_idx = rng.choice(len(x_test), N_TEST_SVM, replace=False)

x_train_svm = x_train[train_idx]
y_train_svm = y_train[train_idx]
x_test_svm = x_test[test_idx]
y_test_svm = y_test[test_idx]

x_train_full_flat = x_train_svm.reshape(N_TRAIN_SVM, -1)
x_test_full_flat = x_test_svm.reshape(N_TEST_SVM, -1)



# 7. SVM klasifikacija na PUNIM slikama
print("\n===== SVM na PUNIM slikama (28x28 = 784 značajke) =====")
svm_full = SVC(kernel="rbf", gamma="scale")
svm_full.fit(x_train_full_flat, y_train_svm)
pred_full = svm_full.predict(x_test_full_flat)

acc_full = accuracy_score(y_test_svm, pred_full)
print(f"Točnost (pune slike): {acc_full:.4f}")
print(classification_report(y_test_svm, pred_full))



# 8. SVM klasifikacija na REDUCIRANIM podacima (Central Layer, 64 broja)
latent_train = encoder.predict(x_train_svm)   # (5000, 64) - već je ravan vektor!
latent_test = encoder.predict(x_test_svm)     # (2000, 64)

print(f"\n===== SVM na REDUCIRANIM podacima - Central Layer "
      f"(dimenzija vektora: {latent_train.shape[1]}) =====")

svm_reduced = SVC(kernel="rbf", gamma="scale")
svm_reduced.fit(latent_train, y_train_svm)
pred_reduced = svm_reduced.predict(latent_test)

acc_reduced = accuracy_score(y_test_svm, pred_reduced)
print(f"Točnost (Central Layer, 64): {acc_reduced:.4f}")
print(classification_report(y_test_svm, pred_reduced))


# 9. Usporedni prikaz SVM točnosti
svm_results = {
    "Pune slike": acc_full,
    "Central Layer (64)": acc_reduced
}

plt.figure(figsize=(6, 5))
plt.bar(svm_results.keys(), svm_results.values(), color=["gray", "salmon"])
plt.ylabel("Točnost (Accuracy)")
plt.title("Usporedba SVM točnosti: pune vs. reducirane slike")
plt.ylim(0, 1)
for i, (k, v) in enumerate(svm_results.items()):
    plt.text(i, v + 0.02, f"{v:.3f}", ha="center")
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()

print("\n===== SAŽETAK TOČNOSTI =====")
for k, v in svm_results.items():
    print(f"{k}: {v:.4f}")
```
